```python
# establish home directory for project
import os
home = os.getcwd()
home
```

```python
# import dataset from roboflow
from roboflow import Roboflow
rf = Roboflow(api_key="")
project = rf.workspace("trail-camera").project("sam-barrett")
version = project.version(10)
dataset = version.download("yolov12")
```

```python
# check ultralytics version
import ultralytics
from ultralytics import YOLO
ultralytics.checks()
```

```python
# train the model
%cd {home}
!yolo task=detect mode=train data={dataset.location}/data.yaml model=yolo12n.pt epochs=100 imgsz=640 device=0 plots=True
```

```python
# python implementation of training
model = YOLO("yolo12n.pt")
results = model.train(data=(f"{dataset.location}/data.yaml"), epochs=100, imgsz=640, device=0, plots=True)
```

```python
# make sure to set to the correct 'train' run directory
%cd {home}
!yolo task=detect mode=val model={home}/runs/detect/train3/weights/best.pt data={dataset.location}/data.yaml
```

```python
# python implementation of validation
model = YOLO(f"{home}/runs/detect/train3/weights/best.pt")
metrics = model.val(data=(f"{dataset.location}/data.yaml"))
```
